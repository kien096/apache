# Thuc hanh 1 - Word Count

Notebook nay chay duoc ngay trong Jupyter.

- Doc file `sample.txt` trong thu muc notebook
- Day du lieu len Spark cluster bang `parallelize`
- Dem so lan xuat hien cua moi tu


```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder.appName('WordCountNotebook')
    .master('spark://master:7077')
    .config('spark.pyspark.python', '/opt/conda/bin/python')
    .getOrCreate()
)

sc = spark.sparkContext
print('Spark master =', sc.master)
```

```python
input_file = '/opt/workspace/notebooks/sample.txt'

with open(input_file, encoding='utf-8') as f:
    lines_list = [line.strip() for line in f if line.strip()]

print('So dong doc duoc:', len(lines_list))
print(lines_list)
```

```python
# parallelize day du lieu tu driver len cluster de worker xu ly.
lines = sc.parallelize(lines_list, 4)
words = lines.flatMap(lambda line: line.split())
pairs = words.map(lambda word: (word, 1))
counts = pairs.reduceByKey(lambda a, b: a + b)
result = counts.sortBy(lambda item: (-item[1], item[0])).collect()

print('WORD COUNT RESULT')
for word, count in result:
    print(f'{word}\t{count}')
```
